```python
import os
import time
import fastf1
import pandas as pd
from fastf1.ergast import Ergast

# 1. Setup
os.makedirs('data', exist_ok=True)
fastf1.Cache.enable_cache('data')
ergast = Ergast()

# 2. Current standings
standings = ergast.get_driver_standings(season=2026).content[0]
print(standings[['position', 'points', 'wins', 'familyName', 'constructorNames']].head(10))

# 3. Races remaining in 2026
schedule = fastf1.get_event_schedule(2026, include_testing=False)
remaining = schedule[schedule['EventDate'] >= pd.Timestamp.now()]
print("\nRaces remaining:", len(remaining))
print(remaining[['RoundNumber', 'EventName', 'EventDate']])

# 4. Download past results 2022-2026
all_results = []
for season in range(2022, 2027):
    sched = fastf1.get_event_schedule(season, include_testing=False)
    past = sched[sched['EventDate'] < pd.Timestamp.now()]
    for rnd in past['RoundNumber']:
        try:
            res = ergast.get_race_results(season=season, round=rnd).content[0]
            res['season'] = season
            res['round'] = rnd
            all_results.append(res)
            print(f"OK {season} round {rnd}")
        except Exception as e:
            print(f"Skip {season} round {rnd}: {e}")
        time.sleep(0.5)

# 5. Combine and save
results = pd.concat(all_results, ignore_index=True)
results.to_csv('data/race_results.csv', index=False)
print("\nDone! Shape:", results.shape)
```

```python
# A. Races per season (check data is complete)
print("RACES PER SEASON")
print(results.groupby('season')['round'].nunique())

# B. Column names
print("\nCOLUMNS")
print(results.columns.tolist())

# C. Grid position of race winners
print("\nWINNERS BY STARTING POSITION")
winners = results[results['position'] == 1]
print(winners['grid'].value_counts().sort_index().head(10))

# D. Points per driver in 2026 (main races only)
print("\n2026 POINTS (RACES ONLY)")
r26 = results[results['season'] == 2026]
print(r26.groupby('familyName')['points'].sum().sort_values(ascending=False).head(10))
```

```python
# Sort races in time order
df = results.sort_values(['season', 'round']).reset_index(drop=True)

# Target: 1 if the driver won
df['won'] = (df['position'] == 1).astype(int)

# Grid 0 means started from pit lane -> treat as last
df['grid'] = df['grid'].replace(0, 22)

# Driver form: average finish in his PREVIOUS 5 races
df['driver_form'] = (df.groupby('driverId')['position']
                       .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean()))

# Team form: average points per car in the team's PREVIOUS 10 entries (= ~5 races x 2 cars)
df['team_form'] = (df.groupby('constructorId')['points']
                     .transform(lambda s: s.shift(1).rolling(10, min_periods=1).mean()))

# Points scored so far this season, before this race
df['driver_season_pts'] = (df.groupby(['season', 'driverId'])['points']
                             .transform(lambda s: s.shift(1).cumsum()))

# First races have no history -> fill with neutral values
df['driver_form'] = df['driver_form'].fillna(12)
df['team_form'] = df['team_form'].fillna(0)
df['driver_season_pts'] = df['driver_season_pts'].fillna(0)

features = ['grid', 'driver_form', 'team_form', 'driver_season_pts']
df[['season', 'round', 'familyName', 'constructorName'] + features + ['won']].tail(22)
```

```python
import os
import fastf1
import pandas as pd
from fastf1.ergast import Ergast
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier

# ---------- LOAD ----------
os.makedirs('data', exist_ok=True)
fastf1.Cache.enable_cache('data')
ergast = Ergast()
standings = ergast.get_driver_standings(season=2026).content[0]
schedule = fastf1.get_event_schedule(2026, include_testing=False)
remaining = schedule[schedule['EventDate'] >= pd.Timestamp.now()]
results = pd.read_csv('data/race_results.csv')

# ---------- FEATURES ----------
df = results.sort_values(['season', 'round']).reset_index(drop=True)
df['won'] = (df['position'] == 1).astype(int)
df['grid'] = df['grid'].replace(0, 22)
df['driver_form'] = df.groupby('driverId')['position'].transform(
    lambda s: s.shift(1).rolling(5, min_periods=1).mean()).fillna(12)
df['team_form'] = df.groupby('constructorId')['points'].transform(
    lambda s: s.shift(1).rolling(10, min_periods=1).mean()).fillna(0)
df['driver_season_pts'] = df.groupby(['season', 'driverId'])['points'].transform(
    lambda s: s.shift(1).cumsum()).fillna(0)
features = ['grid', 'driver_form', 'team_form', 'driver_season_pts']

# ---------- TRAIN & TEST ----------
train = df[df['season'] <= 2025]
test = df[df['season'] == 2026].copy()

def score(name, probs):
    test['prob'] = probs
    picks = test.loc[test.groupby('round')['prob'].idxmax()]
    print(f"{name}: picked the winner in {picks['won'].sum()} of {len(picks)} races")

score("Baseline (pole wins)", -test['grid'])

lr = LogisticRegression(max_iter=1000).fit(train[features], train['won'])
score("Logistic Regression", lr.predict_proba(test[features])[:, 1])

gb = GradientBoostingClassifier(random_state=42).fit(train[features], train['won'])
score("Gradient Boosting", gb.predict_proba(test[features])[:, 1])
```

simulate the championship

Two problems to solve:

We don’t know the future grid positions. So for each driver we use his average grid from his last 5 races as an estimate.
We need full results, not just the winner, because P2, P3… also score points. Trick: imagine a hat with tickets. Favourites have more tickets. Draw the winner, then draw P2 from the remaining tickets, and so on. That gives a realistic random race result. (The math name is the Plackett-Luce model, a nice thing to mention in your project.)

```python
import numpy as np

# ---- 1. Each driver's current strength ----
r26 = df[df['season'] == 2026]
last_round = r26['round'].max()
current = r26[r26['round'] == last_round][['driverId', 'familyName', 'constructorId']]

rows = []
for _, d in current.iterrows():
    hist = df[df['driverId'] == d.driverId].tail(5)
    team = df[df['constructorId'] == d.constructorId].tail(10)
    rows.append({
        'driverId': d.driverId,
        'familyName': d.familyName,
        'grid': hist['grid'].mean(),
        'driver_form': hist['position'].mean(),
        'team_form': team['points'].mean(),
        'driver_season_pts': r26[r26['driverId'] == d.driverId]['points'].sum(),
    })
nxt = pd.DataFrame(rows)

nxt['win_prob'] = gb.predict_proba(nxt[features])[:, 1]
nxt['win_prob'] /= nxt['win_prob'].sum()     # make probabilities add up to 100%

print("NEXT RACE: WIN PROBABILITY")
print(nxt.sort_values('win_prob', ascending=False)[['familyName', 'grid', 'driver_form', 'win_prob']].head(8))

# ---- 2. Simulate the rest of the season ----
POINTS = np.array([25, 18, 15, 12, 10, 8, 6, 4, 2, 1] + [0] * (len(nxt) - 10))
start = nxt['driverId'].map(standings.set_index('driverId')['points']).fillna(0).values.astype(float)
logp = np.log(np.clip(nxt['win_prob'].values, 1e-6, None))

n_sims = 10000
n_races = len(remaining)
rng = np.random.default_rng(42)
champs = np.zeros(len(nxt))

for _ in range(n_sims):
    total = start.copy()
    for _ in range(n_races):
        order = np.argsort(-(logp + rng.gumbel(size=len(nxt))))   # the "hat draw"
        total[order] += POINTS
    champs[np.argmax(total)] += 1

nxt['title_%'] = 100 * champs / n_sims

print("\n2026 WORLD CHAMPIONSHIP: CHANCE TO WIN")
print(nxt.sort_values('title_%', ascending=False)[['familyName', 'win_prob', 'title_%']].head(8))
```

```python
def simulate(temp=1.0, dnf_rate=0.0, n_sims=10000):
    rng = np.random.default_rng(42)
    lp = logp / temp
    champs = np.zeros(len(nxt))
    for _ in range(n_sims):
        total = start.copy()
        for _ in range(n_races):
            order = np.argsort(-(lp + rng.gumbel(size=len(nxt))))
            dnf = rng.random(len(nxt)) < dnf_rate
            order = np.concatenate([order[~dnf[order]], order[dnf[order]]])  # DNFs go to the back
            total[order] += POINTS
        champs[np.argmax(total)] += 1
    return (pd.Series(100 * champs / n_sims, index=nxt['familyName'])
              .sort_values(ascending=False).head(5))

for temp, dnf in [(1, 0), (1, 0.08), (3, 0.08)]:
    print(f"\nTemperature={temp}, DNF rate={dnf}")
    print(simulate(temp, dnf))
```
